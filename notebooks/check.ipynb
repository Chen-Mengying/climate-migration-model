```python
import pandas as pd

# === 通用数据检查函数 ===
def check_data(df, name):
    print(f"\n=== 检查数据集: {name} ===")
    print(f"数据集大小: {df.shape}")
    print("\n数据类型:")
    print(df.dtypes)
    print("\n空值统计:")
    print(df.isnull().sum())
    print("\n前 5 行数据:")
    print(df.head())
    print("\n基本统计信息:")
    print(df.describe(include='all'))
```

```python
# === 读取数据集 ===
# 示例：迁移流数据
migration_path = '../data/processed/migration_flow_cleaned.csv'
migration_df = pd.read_csv(migration_path)
check_data(migration_df, "迁移流数据")
```

```python


# 示例：WDI 控制变量
wdi_path = '../data/processed/control_wdi_variables.csv'
wdi_control_vars = pd.read_csv(wdi_path)
check_data(wdi_control_vars, "WDI 控制变量")

# 示例：地理距离数据
dist_path = '../data/processed/control_geodistance.csv'
dist_df = pd.read_csv(dist_path)
check_data(dist_df, "地理距离数据")

# 示例：气候数据
spei_path = '../data/processed/migration_with_spei.csv'
spei_df = pd.read_csv(spei_path)
check_data(spei_df, "气候数据")

# 示例：边界摩擦数据
border_path = '../data/processed/border_friction_panel_ew.csv'
border_friction_df = pd.read_csv(border_path)
check_data(border_friction_df, "边界摩擦数据")
```
